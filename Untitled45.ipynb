## 1 Скачивание словаря и корпуса
Код ниже предназначен для получения словаря и корпуса OpenCorpora.

Словарь используется для того чтобы:
*   по словоформе получить все возможные леммы
*   нахождения инфинитивов для лемм.

Корпус используется для:


*  Проверки тегов на реальных примерах с контекстом
*  Оценки качества лемматизации (сравнение с правильными ответами)




````python
DICT_FN = 'dict.opcorpora.xml'
DICT_BZ = DICT_FN + '.bz2'
CORPUS_FN = 'annot.opcorpora.no_ambig.xml'
CORPUS_BZ = CORPUS_FN + '.bz2'
DICT_URL = 'http://opencorpora.org/files/export/dict/dict.opcorpora.xml.bz2'
CORPUS_URL = 'http://opencorpora.org/files/export/annot/annot.opcorpora.no_ambig.xml.bz2'


def download_file(url, filename):
    if os.path.exists(filename):
        print(f"{filename} уже существует.")
        return
    print(f"Скачиваем {filename} ...")
    with requests.get(url, stream=True) as r: # не загружаем весь файл в оперативную память целиком, а читаем его потоково
        r.raise_for_status() #успешен ли запрос
        with open(filename, 'wb') as f:
            for chunk in r.iter_content(chunk_size=1024 * 1024): #за один шаг цикла будет обработан 1мб данных
                if chunk:
                    f.write(chunk)
    print(f"Файл: {filename}")

download_file(DICT_URL, DICT_BZ)
download_file(CORPUS_URL, CORPUS_BZ)


def unpack(bz_path, out_path):
    if os.path.exists(out_path):
        print(f"{out_path} уже существует.")
        return
    print(f"Распаковываем {bz_path} -> {out_path} ...")
    with bz2.open(bz_path, 'rb') as f_in, open(out_path, 'wb') as f_out:
        f_out.write(f_in.read())
    print("ОК.")

unpack(DICT_BZ, DICT_FN)
unpack(CORPUS_BZ, CORPUS_FN)
````

## Нормализация и преобразование частей речи

````python
def normalize_word(w: str) -> str:
    return w.lower().replace('ё', 'е')

#отображаем части речи OpenCorpra в упрощенные теги
gram_to_tag = {
    'NOUN': 'S', 'NPRO': 'NI',
    'ADJF': 'A', 'ADJS': 'A', 'COMP': 'A',
    'PRTF': 'V', 'PRTS': 'V',
    'VERB': 'V', 'INFN': 'V', 'GRND': 'V',
    'CONJ': 'CONJ',
    'PREP': 'PR',
    'ADVB': 'ADV', 'PRED': 'ADV',
    'PRCL': 'ADV', 'INTJ': 'ADV',
    'NUMR': 'A', 'NUMB': 'A',
    'PNCT': '',
}
````

````python
print("Парсинг словаря...")

tree = ET.parse(DICT_FN)
root = tree.getroot()

# form -> list of (head_lemma, form_tag, grams)
form_to_lemmas = {}

# head -> list of (form_string, grams)  (обратный индекс)
head2forms = {}

# lemma id mappings (для aspect links - сохраняем только для совместимости)
lemma_id_to_norm = {}
lemma_id_to_grams = {}

# lemma -> best INFN form (string) if detected in paradigm
lemma_to_inf = {}

# lemma -> INFN when lemma itself marked impf (try prefer)
lemma_to_impf = {}

for lemma_node in root.findall('.//lemma'):
    lid = lemma_node.get('id')
    l = lemma_node.find('l')
    if l is None:
        continue
    head_raw = l.get('t')
    head = normalize_word(head_raw)
    grams_head = [g.get('v') for g in l.findall('g')]

    lemma_id_to_norm[lid] = head
    lemma_id_to_grams[lid] = grams_head

    pos_head = grams_head[0] if grams_head else 'UNK'
    tag_head = gram_to_tag.get(pos_head, 'UNK')

    # init
    head2forms.setdefault(head, [])
    # include head itself as a form
    head2forms[head].append((head, grams_head))
    form_to_lemmas.setdefault(head, [])
    entry_head = (head, tag_head, grams_head)
    if entry_head not in form_to_lemmas[head]:
        form_to_lemmas[head].append(entry_head)

    # detect INFN in head
    inf_found = None
    if 'INFN' in grams_head:
        inf_found = head

    # detect aspect of head
    head_aspect = None
    if 'perf' in grams_head:
        head_aspect = 'perf'
    if 'impf' in grams_head:
        head_aspect = 'impf'

    # forms
    for f in lemma_node.findall('f'):
        f_text_raw = f.get('t')
        f_text = normalize_word(f_text_raw)
        f_grams = [g.get('v') for g in f.findall('g')]
        f_pos = f_grams[0] if f_grams else pos_head
        f_tag = gram_to_tag.get(f_pos, tag_head)

        form_to_lemmas.setdefault(f_text, [])
        entry = (head, f_tag, f_grams)
        if entry not in form_to_lemmas[f_text]:
            form_to_lemmas[f_text].append(entry)

        head2forms.setdefault(head, [])
        head2forms[head].append((f_text, f_grams))

        # INFN detection among forms
        if 'INFN' in f_grams:
            inf_found = f_text

    if inf_found:
        lemma_to_inf[head] = inf_found
        if head_aspect == 'impf':
            lemma_to_impf[head] = inf_found

# Убираем построение comp_map и aspect_pairs полностью

print(f"Словарь загружен: {len(form_to_lemmas):,} форм")
print(f"lemma->INFN: {len(lemma_to_inf):,}")
print(f"lemma impf: {len(lemma_to_impf):,}")

# build set of all known INFN forms (for heuristics)
known_inf_forms = set(lemma_to_inf.values())
````

## Парсинг корпусов

Блок кода ниже предназначен для парсинга корпусов. Рассмотрим его работу подробнее.В начале происходит парсинг XML-файла и создаётся объект дерева, далее
```root2 = tree2.getroot()```
Получает корневой элемент дерева, от которого можно искать все остальные элементы. Создаётся пустой список sentences, в который будут сохраняться все предложения из корпуса в виде списков токенов.

После этого цикл по всем предложениям: ищем все предлодения в дереве, для каждого найденного предложения ```(sent_node)``` выполняется следующее: Создаётся пустой список ```toks```, в который будут сохраняться токены текущего предложения, далее цикл по всем токенам предложения: ищем все элементы ```<token>``` внутри ```<tokens>``` для текущего предложения.
Для каждого токена выполняется дальнейшая обработка: ```token.get('text')```: получаем текст токена (например, "Кот").
```if w in '.,?!'```: continue: пропускаем знаки препинания.

Далее ```token.find('./tfr/v')``` Ищет элемент <v> внутри <tfr> для текущего токена. <v> содержит информацию о лемме и грамматических тегах.
```if v is None```: continue: Если разметки нет, токен пропускается.
```
<v>: Вариант разметки. Содержит один или несколько элементов <l> (лемм).
<l>: Лемма токена. Имеет атрибут t (текст леммы) и список грамматических тегов (<g>).
```

Далее происходит извлечение леммы токена: ```v.find('l')```: Ищет элемент <l> внутри <v>. <l> содержит лемму (t) и грамматические теги (g).
```if l is None: continue``` -Если леммы нет, токен пропускается.
```lemma_raw = l.get('t')```: Получает текст леммы (например, "кот" для токена "Кот").
Также происходит извлечение грамматических тегов: ```[g.get('v') for g in l.findall('g')]```: Извлекает все грамматические теги для токена.
```pos = grams[0] if grams else 'UNK'```: Берёт первый тег (часть речи) или 'UNK' (unknown), если тегов нет.
```tag = gram_to_tag.get(pos, 'UNK')```: Преобразует OpenCorpora-тег в упрощённый тег (например, "NOUN" -> "S").

Добавляем кортеж в список toks:

```w``` - Исходный текст токена (например, "Кот").

```normalize_word(w)``` - Нормализованный текст токена (например, "кот").

```normalize_word(lemma_raw)``` - Нормализованная лемма (например, "кот").

```tag``` - Упрощённый POS-тег (например, "S").

После этого если список ```toks``` не пуст добавляем его в список ```sentences```


````python
print("Парсинг корпуса...")
tree2 = ET.parse(CORPUS_FN) #загружаем корпус
root2 = tree2.getroot()  #получаем корневой элемент

sentences = []
#цикл по всем предложениям в дереве
for sent_node in root2.findall('.//sentence'):
    toks = []
    for token in sent_node.findall('./tokens/token'):
        w = token.get('text')
        if w in '.,?!':
            continue
        v = token.find('./tfr/v')
        if v is None:
            continue
        l = v.find('l')
        if l is None:
            continue
        lemma_raw = l.get('t')
        grams = [g.get('v') for g in l.findall('g')]
        pos = grams[0] if grams else 'UNK'
        tag = gram_to_tag.get(pos, 'UNK')
        toks.append((w, normalize_word(w), normalize_word(lemma_raw), tag))
    if toks:
        sentences.append(toks)

print(f"Корпус: {len(sentences):,} предложений")

if len(sentences) == 0:
    raise RuntimeError("Корпус пуст!")

#разбиваем список предложений на обучающию и тестовую выборку
train_sents, test_sents = train_test_split(sentences, test_size=0.2, random_state=42)
````

## Извлечение признаков
Ячейка кода ниже предназначена для извлечения признаков для модели "Случайный лес"

Функция получает cписок токенов в предложении. Каждый токен — это кортеж:
```python
(исходное_слово, нормализованное_слово, нормализованная_лемма, POS-тег)
```
Например: ```[("Кот", "кот", "кот", "S"), ("съел", "съел", "съесть", "V")]```

А также i - Индекс текущего токена в предложении.

prev_tag - тег предыдущего токена.


Функция  преобразует каждый токен в набор информативных признаков, которые позволяют модели POS-tagging учитывать:


*  Форму слова (префиксы, суффиксы).
*  Контекст (соседние слова и их теги).
*  Длину





````python
def extract_features(sentence, i, prev_tag=None):
    w = sentence[i][1] # получаем нормализованное слово текущего токена
    feats = {'len': len(w)} #словарь признаков: первый признак - длина слова
    #извлекаем преф и суф
    for k in range(1, 5):
        feats[f'pref{k}'] = w[:k]
        feats[f'suf{k}'] = w[-k:]
    #добавление тега предыдущ слова
    if prev_tag:
        feats['prev_tag'] = prev_tag
    #если текущий токен не первый в предложении (i > 0), извлекаются последние 3 символа предыдущего токена
    if i > 0:
        feats['prev3'] = sentence[i-1][1][-3:]
    #Если текущий токен не последний в предложении (i < len(sentence) - 1), извлекаются первые 3 символа следующего токена
    if i < len(sentence) - 1:
        feats['next3'] = sentence[i+1][1][:3]
    return feats
````

## Подготовка данных для модели и ее обучение

Блок кода ниже предназначен для построения обучающих и тестовых данных.

```X_train```: Список, в который будут сохраняться признаки (features) для каждого токена.

```y_train```: Список, в который будут сохраняться истинные теги (метки классов) для каждого токена.

Цикл по всем прдложениям в обучающей выборке можно описать так:

```for sent in train_sents```:

Проходим по каждому предложению в обучающей выборке .
(Каждое предложение — это список токенов, где каждый токен представлен кортежем)
```
(исходное_слово, нормализованное_слово, нормализованная_лемма, POS-тег)
```

```prev = 'START'```

Инициализируем prev (тег предыдущего токена) специальным значением 'START'.
Это нужно, чтобы для первого токена в предложении передавать какой-то "псевдо-тег" (так как у первого токена нет предыдущего).


```for i in range(len(sent))```:

Проходим по каждому токену в предложении.


```X_train.append(extract_features(sent, i, prev))```

Для текущего токена (sent[i]) извлекаем признаки с помощью функции extract_features.
Добавляем эти признаки в X_train.


```y_train.append(sent[i][3])```

Добавляем истинный тег текущего токена (sent[i][3]) в y_train.


```prev = sent[i][3]```

Обновляем prev — теперь он равен тегу текущего токена.
Это нужно для извлечения признаков следующего токена (чтобы передать тег предыдущего токена).

Далее происходит обучение модели и оценивание качества, побробнее можно почитать из комметариев.


````python
X_train, y_train = [], []
for sent in train_sents:
    prev = 'START'
    for i in range(len(sent)):
        X_train.append(extract_features(sent, i, prev))
        y_train.append(sent[i][3])
        prev = sent[i][3]

vectorizer = DictVectorizer(sparse=True) #векторизуем признаки
X_train_vec = vectorizer.fit_transform(X_train) #обучаем векторизатор на наших обучающих данных

print("Обучение RandomForest...")
clf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
clf.fit(X_train_vec, y_train) #обучаем модель на векторизованных признаках  (Модель учится предсказывать тег (y_train) по признакам токена (X_train_vec)

# оценка
X_test, y_test, y_test_lemmas = [], [], []

#проходим по всем токенам в тестовых предложениях (test_sents)
for sent in test_sents:
    #Для каждого токена
    for i in range(len(sent)):
        # извлекаем признаки, добавл их в X_test
        X_test.append(extract_features(sent, i))
        # добавляем тег в y_test
        y_test.append(sent[i][3])
        # добавляем лемму
        y_test_lemmas.append(sent[i][2])


X_test_vec = vectorizer.transform(X_test) # преобразуем тестовые признаки
y_pred = clf.predict(X_test_vec) # предсказания модели
pos_acc = accuracy_score(y_test, y_pred)
print(f"POS-accuracy на тесте: {pos_acc:.4f}")

````

##  Поиск инфинитив для глагола
Данная функция выполняет несколько типов поиска инфинитива, рассмотрим каждый из них детальнее:

Берёт все формы (forms) для леммы из словаря head2forms.
Ищет форму, у которой в грамматических тегах (grams) есть и INFN (инфинитив), и impf (несовершенный вид). Это самый приоритетный источник поиска, так как сразу даёт инфинитив несовершенного вида.

Далее, если не найден инфинитив несовершенного вида, ищет любой инфинитив (даже совершенного вида).

Далее осуществляется поиск в словарях ```lemma_to_impf и lemma_to_inf``` они заполняются на этапе парсинга словаря OpenCorpora.

````python
def find_inf_for_head(head):

    forms = head2forms.get(head, [])

    for form, grams in forms:
        if isinstance(grams, list) and 'INFN' in grams and 'impf' in grams:
            return form, grams

    for form, grams in forms:
        if isinstance(grams, list) and 'INFN' in grams:
            return form, grams

    if head in lemma_to_impf:
        inf = lemma_to_impf[head]
        grams = form_to_lemmas.get(inf, [(inf, None, [])])[0][2] if form_to_lemmas.get(inf) else []
        return inf, grams

    if head in lemma_to_inf:
        inf = lemma_to_inf[head]
        grams = form_to_lemmas.get(inf, [(inf, None, [])])[0][2] if form_to_lemmas.get(inf) else []
        return inf, grams

    return None, []
````

## Построение инфинитива, в случае если он не найден

Блок кода ниже используется как "резервный метод", когда стандартные способы поиска инфинитива не сработали

Функция synthesize_inf_from_token() пытается построить правдоподобные кандидаты на роль инфинитива из строки токена. Она принимает нормализированный токен и возвращает инфинитив.

Рассмторим работу данной функции подробнее. В начале происходит удаление постфикса ```(ся, сь)```. Проверяет, оканчивается ли токен на -ся или -сь.
Если да, удаляет эти суффиксы и запоминает, что глагол возвратный.
Если нет, оставляет токен без изменений

Далее происходит удаление типичных окончаний для получения основы, это работает следующим образом: cоздается множество ```stems```, в которое добавляются возможные основы слова, пробуем удалить из t_core типичные окончания (из списка strips), если после удаления остаётся не менее 2 символов.

Список ```strips```:  содержит окончания прошедшего времени (-л, -ла, -ло), причастий (-вший, -вшийся), деепричастий (-я, -а), личных окончаний глаголов (-ю, -ешь, -ут)

Далее происходит удаление 1–4 последних символов для получения основ. это осуществляется следующим образом: lобавляет в ```stems``` основы, полученные удалением 1, 2, 3 или 4 последних символов из ```t_core```.
Проверяет, что после удаления остаётся не менее 2 символов.

Пример:
```
t_core = "читал" -> удаляем 1 символ -> "чита", 2 символа -> "чит", 3 символа -> "чи", 4 символа  "ч" (но len("ч") == 1 < 2, поэтому не добавляется).
```

Последним шагом является генерация кандидатов на инфинитив, рассмторим работу более детально: проходим по всем основам ```(stems)```, отсортированным по убыванию длины (чтобы сначала пробовать более длинные основы).
Для каждой основы ```(stem)``` добавляем окончания инфинитива (INF_ENDINGS).
Проверяем, существует ли полученный кандидат ```(cand)``` в множестве известных инфинитивов ```(known_inf_forms)``` - словарь, который формируется на этапе парсинга OpenCorpra.
Если токен был возвратным ```(reflexive = True)```, также проверяет кандидат с постфиксом.

Если же ни один из сгенерированных кандидатов не найден в ```known_inf_forms```, возвращаем None






````python
#INF_ENDINGS = ['ть', 'ти', 'ить', 'ать', 'еть', 'уть', 'ыть', 'ться', 'тись']  # try candidates
INF_ENDINGS = ['ть', 'ти']

IRREGULAR_VERBS = {
    'вывел': 'вывести',
    'вывели': 'вывести',
    'вывела': 'вывести',
    'вывело': 'вывести',
    'пошёл': 'пойти',
    'пошла': 'пойти',
    'пошло': 'пойти',
    'пошли': 'пойти',
    'пришёл': 'прийти',
    'пришла': 'прийти',
    'пришло': 'прийти',
    'пришли': 'прийти',
    'нашёл': 'найти',
    'нашла': 'найти',
    'нашло': 'найти',
    'нашли': 'найти',
    'ушел': 'уйти',
    'ушла': 'уйти',
    'ушло': 'уйти',
    'ушли': 'уйти',
}

def synthesize_inf_from_token(token_norm):

    if token_norm in IRREGULAR_VERBS:
        return IRREGULAR_VERBS[token_norm]

    t = token_norm

    reflexive = False
    if t.endswith('ся'):
        t_core = t[:-2]
        reflexive = True
    elif t.endswith('сь'):
        t_core = t[:-2]
        reflexive = True
    else:
        t_core = t
    strips = ['авши', 'явши', 'ивши', 'ившись', 'явшись', 'авшись',
              'шись', 'ывшись', 'ясь', 'ась', 'или', 'ыла', 'ыло',
              'ила', 'или', 'ал', 'ала', 'али', 'ло', 'ли', 'л',
              'ю', 'ешь', 'ете', 'ут', 'ют', 'ит', 'ат', 'ят',
              'ем', 'ешь', 'ешься', 'лся', 'лись']
    stems = set()
    stems.add(t_core)
    for s in strips:
        if t_core.endswith(s) and len(t_core) - len(s) >= 2:
            stems.add(t_core[:-len(s)])
    for k in range(1, 5):
        if len(t_core) - k >= 2:
            stems.add(t_core[:-k])

    for stem in sorted(stems, key=lambda x: -len(x)):
        for end in INF_ENDINGS:
            cand = stem + end
            if reflexive:
                cand_r = cand + 'ся'
                if cand_r in known_inf_forms:
                    return cand_r
            if cand in known_inf_forms:
                return cand
    return None
````

## Выбор лучшей леммы

Функция ```choose_best_lemma()``` выбирает наиболее подходящую лемму для токена на основе его тега и возможных лемм. Она использует специальные правила для глаголов, существительных и прилагательных.
Она принимает ```possible```: список возможных лемм для токена (каждый элемент — кортеж (head, form_tag, grams)).

```tag```: тег токена (например, 'V' для глагола, 'S' для существительного).

```norm_form```: нормализованная форма токена.

Возвращает: ```(lemma, grams)```: выбранная лемма и её грамматические теги.

Рассмотрим работу функции подробнее: если список возможных лемм пуст, возвращает (None, []). Далее используется вспомогательная функция ```has_flag()```: они принимает список грамматических тегов и флаг, наличие которого надо проверить, возвращает ```True/False``` в зависимости от наличия флага.

В начале извлекаем первую лемму (base_head) и её грамматические теги (base_grams) из списка ```possible```.
```base_head``` используется как "резервный вариант", если не удастся найти более подходящую лемму. Далее собираем уникальные леммы из списка possible.
Далее проверяем является ли токен глаголом, причастием или деепричастием, выполняется проверка есть ли среди грамматических тегов токена теги:
```
'PRTF' (причастие),
'PRTS' (краткое причастие),
'GRND' (деепричастие)
```

Используется проверка наличия хотя бы одного из этих тегов в любом наборе грамматических тегов ```(g) из possible```.

Далее treat_as_verb: устанавливается в True, если тег токена — глагол ```(tag == 'V')```. Или токен — причастие/деепричастие ```(token_has_prtf == True).```

Далее происходит обработка таких слов, рассмотрим более детально: осуществляется поиск инфинитива несовершенного вида для каждой леммы-кандидата, это работает таким образом: для каждой леммы (head) из ```candidate_heads``` вызывается ```find_inf_for_head```, которая ищет инфинитив несовершенного вида.
Если инфинитив найден, возвращаем  (```inf, grams)```.

Также ищем инфинитив среди возможных лемм токена дял этого проверяем, есть ли среди возможных лемм токена такая, у которой есть тег ```'INFN'```. Если да, возвращает эту лемму и её грамматические теги.

Следующий шаг - получение инфинтива из токена, пытаемся синтезировать инфинитив из нормализованной формы токена с помощью функции ```synthesize_inf_from_token()```.
Если синтез успешен, возвращает ```(synth, grams_val)```.

Далее проверяем, есть ли базовая лемма в словаре ```lemma_to_inf```.
Если да, возвращаем инфинитив из lemma_to_inf и его грамматические теги.

В завершении пытаемся найти инфинитив для базовой леммы с помощью ```find_inf_for_head```.
Если инфинитив найден, возвращает его.
Если нет, возвращаем базовую лемму и её грамматические теги

Вернемся теперь к описанию основной функции: для каждого кандидата ищем форму с тегами 'NOMN' (именительный падеж) и 'sing' (единственное число).
Если найдена, возвращаем лемму и часть речи. Далее проиходит обработка прилагательных рассмотрим детально: сначала ищем форму с тегами 'NOMN', 'sing' и 'masc' (мужской род).
Если не найдена, ищем форму с 'NOMN' и 'sing' (без указания рода).
Если не найдена, возвращаем ```(base_head, base_grams)```.
Для всех остальных тегов возвращаем базовую лемму.














````python
def choose_best_lemma(possible, tag, norm_form=None):

    if not possible:
        return None, []

    def has_flag(grams, flag):
        return isinstance(grams, list) and flag in grams

    # baseline
    base_head, _, base_grams = possible[0]

    # cобираем уникальные леммы
    candidate_heads = []
    for h, _, _ in possible:
        if h not in candidate_heads:
            candidate_heads.append(h)

    #проверяем, есть ли среди грамматических тегов токена теги причастие краткое причастие деепричастие
    token_has_prtf = any(isinstance(g, list) and any(x in g for x in ('PRTF', 'PRTS', 'GRND')) for _, _, g in possible)
    treat_as_verb = (tag == 'V') or token_has_prtf

    if treat_as_verb:
        for head in candidate_heads:
            inf, grams = find_inf_for_head(head)
            if inf:
                return inf, grams if grams else []

        for h, ftag, grams in possible:
            if has_flag(grams, 'INFN'):
                return h, grams

        synth = synthesize_inf_from_token(norm_form) if norm_form else None
        if synth:
            grams = form_to_lemmas.get(synth, [])
            grams_val = grams[0][2] if grams else []
            return synth, grams_val

        if base_head in lemma_to_inf:
            inf = lemma_to_inf[base_head]
            grams = form_to_lemmas.get(inf, [])
            grams_val = grams[0][2] if grams else []
            return inf, grams_val

        inf, grams = find_inf_for_head(base_head)
        if inf:
            return inf, grams
        return base_head, base_grams

    if tag == 'S':
        for head in candidate_heads:
            for _, _, grams in form_to_lemmas.get(head, []):
                if has_flag(grams, 'NOMN') and has_flag(grams, 'sing'):
                    return head, grams
        return base_head, base_grams

    if tag == 'A':
        for head in candidate_heads:
            for _, _, grams in form_to_lemmas.get(head, []):
                if has_flag(grams, 'NOMN') and has_flag(grams, 'sing') and has_flag(grams, 'masc'):
                    return head, grams
        for head in candidate_heads:
            for _, _, grams in form_to_lemmas.get(head, []):
                if has_flag(grams, 'NOMN') and has_flag(grams, 'sing'):
                    return head, grams
        return base_head, base_grams

    return base_head, base_grams
````

## Лемматизация на тестовой выборке и оценка качества

Блок кода ниже отвечает за лемматизацию тестовой выборки и оценку точности лемматизации. Применяет модель к тестовой выборке, выбирает леммы для каждого токена с помощью функции ```choose_best_lemma```, оценивает точность лемматизации с помощью ```accuracy_score```.

````python
pred_lemmas = [] #список, в который будут сохраняться предсказанные леммы для всех токенов тестовой выборки
idx = 0
for sent in test_sents: #цикл по предложениям тестовой выборки
    for i in range(len(sent)):
        norm = sent[i][1]
        predicted_tag = y_pred[idx]
        possible = form_to_lemmas.get(norm, [])
        lemma, grams = choose_best_lemma(possible, predicted_tag, norm_form=norm) #выбор лучшей леммы
        if lemma is None:
            lemma = norm
            grams = []
        pred_lemmas.append(lemma)
        idx += 1

lem_acc = accuracy_score(y_test_lemmas, pred_lemmas)
print(f"Lemmatization-accuracy на тесте: {lem_acc:.4f}")
````

## Основная функция обработки текста

Функция process_text выполняет полный цикл обработки текста:

Очистка и токенизация,
извлечение признаков,
предсказание тегов,
лемматизация.



````python
def process_text(input_text: str) -> str:
    lines = input_text.strip().split('\n')
    outputs = []
    for line in lines:
        clean = re.sub(r'[.,?!]', '', line).strip()
        if clean == '':
            outputs.append('')
            continue
        tokens = clean.split()
        norm_tokens = [normalize_word(t) for t in tokens]
        sent = [(tokens[i], norm_tokens[i], None, None) for i in range(len(tokens))]
        X = [extract_features(sent, i) for i in range(len(sent))]
        X_vec = vectorizer.transform(X)
        pred_tags = clf.predict(X_vec)
        res = []

        for i, token in enumerate(tokens):
            norm = norm_tokens[i]
            tag = pred_tags[i]
            possible = form_to_lemmas.get(norm, [])
            lemma, grams = choose_best_lemma(possible, tag, norm_form=norm)
            if lemma is None:
                lemma = norm
                grams = []
            res.append(f"{token}{{{lemma}={tag}}}")
        outputs.append(" ".join(res))
    return "\n".join(outputs)
````
